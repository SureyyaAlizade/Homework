```python
score = int(input('Enter score: '))
if 90<score<=100:
    print('Your grade is A')
elif score>=81:
    print('Your grade is B')
elif score>=71:
    print('Your grade is C')
elif score>=61:
    print('Your grade is D')
elif score>=51:
    print('Your grade is E')
else: print('Your grade is F')
```

```python
n = int(input('Enter n: (n>0)'))
sum_of_all = 0
div_by_3 = 0
for i in range(1,n+1):
    if i%5 == 0 and i%3 == 0:
        print(f'{i} -> 3 5')
        div_by_3 += 1
    elif i%3 == 0:
        print(f'{i} -> 3')
        div_by_3 += 1
    elif i%5 == 0:
        print(f'{i} -> 5')
    else: print(i)

    if i%2 == 0:
        sum_of_all += i

print(f'Sum of all even numbers: {sum_of_all}')
print(f'Count of numbers divisible by 3: {div_by_3}')


    


```
